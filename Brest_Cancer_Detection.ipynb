```python
# Import the required libraries

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.datasets import load_breast_cancer

print("Libraries imported successfully!")
```

```python
# Load the Wisconsin Breast Cancer dataset

dataset = load_breast_cancer()

# Convert the features into a DataFrame
df = pd.DataFrame(
    dataset.data,
    columns=dataset.feature_names
)

# Add the diagnosis column
df["diagnosis"] = dataset.target

# Display the first five rows
print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)

df.head()
```

```python
# Display the diagnosis labels

print("Target labels:", dataset.target_names)

print("\nDiagnosis counts:")
print(df["diagnosis"].value_counts().sort_index())

# Convert numerical labels to readable names
df["diagnosis_name"] = df["diagnosis"].map({
    0: "Malignant",
    1: "Benign"
})

print("\nDiagnosis distribution:")
print(df["diagnosis_name"].value_counts())
```

```python
# Check dataset information

print("Dataset information:")
df.info()

print("\nMissing values:")
print(df.isnull().sum().sum())

print("\nDuplicate rows:")
print(df.duplicated().sum())

print("\nStatistical summary:")
display(df.describe().round(2))
```

```python
# Step 1: Diagnosis distribution

plt.figure(figsize=(7, 5))

sns.countplot(
    data=df,
    x="diagnosis_name"
)

plt.title("Distribution of Benign and Malignant Cases")
plt.xlabel("Diagnosis")
plt.ylabel("Number of Patients")
plt.show()

print(df["diagnosis_name"].value_counts())
```

```python
# Step 2: Feature distributions

features_to_plot = [
    "mean radius",
    "mean texture",
    "mean perimeter",
    "mean area",
    "mean smoothness"
]

df[features_to_plot].hist(
    figsize=(12, 8),
    bins=20,
    edgecolor="black"
)

plt.suptitle("Distribution of Selected Breast Cancer Features")
plt.tight_layout()
plt.show()
```

```python
# Step 3: Correlation heatmap

correlation = df[dataset.feature_names].corr()

plt.figure(figsize=(16, 12))

sns.heatmap(
    correlation,
    cmap="coolwarm",
    center=0,
    xticklabels=True,
    yticklabels=True
)

plt.title("Correlation Heatmap of Breast Cancer Features")
plt.xticks(rotation=90, fontsize=7)
plt.yticks(fontsize=7)
plt.tight_layout()
plt.show()
```

```python
# Step 4: Compare feature values by diagnosis

plt.figure(figsize=(8, 5))

sns.boxplot(
    data=df,
    x="diagnosis_name",
    y="mean radius"
)

plt.title("Mean Radius by Diagnosis")
plt.xlabel("Diagnosis")
plt.ylabel("Mean Radius")
plt.show()
```

```python
# Step 5: Average feature comparison

selected_features = [
    "mean radius",
    "mean texture",
    "mean perimeter",
    "mean area",
    "mean smoothness"
]

average_values = df.groupby("diagnosis_name")[
    selected_features
].mean()

display(average_values.round(3))
```

```python
# Step 1: Separate features and target

X = df[dataset.feature_names].copy()
y = dataset.target.copy()

print("Feature shape:", X.shape)
print("Target shape:", y.shape)
print("Feature names:")
print(X.columns.tolist())
```

```python
# Step 2: Split the dataset

from sklearn.model_selection import train_test_split

# Separate 20% for final testing
X_temp, X_test, y_temp, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

# Split the remaining 80% into training and validation
X_train, X_val, y_train, y_val = train_test_split(
    X_temp,
    y_temp,
    test_size=0.25,
    random_state=42,
    stratify=y_temp
)

print("Training set:", X_train.shape)
print("Validation set:", X_val.shape)
print("Testing set:", X_test.shape)
```

```python
# Step 3: Check class distribution

split_data = {
    "Training": y_train,
    "Validation": y_val,
    "Testing": y_test
}

for split_name, labels in split_data.items():
    print(f"\n{split_name} set:")
    print(
        pd.Series(labels)
        .map({0: "Malignant", 1: "Benign"})
        .value_counts()
    )
```

```python
# Step 4: Import preprocessing tools

from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline

print("Preprocessing tools imported successfully!")
```

```python
# Step 5: Prepare the Logistic Regression pipeline

from sklearn.linear_model import LogisticRegression

logistic_pipeline = Pipeline([
    ("scaler", StandardScaler()),
    ("model", LogisticRegression(max_iter=2000, random_state=42))
])

print("Logistic Regression pipeline created!")
print(logistic_pipeline)
```

```python
# Step 1: Import model training and evaluation tools

from sklearn.model_selection import GridSearchCV, StratifiedKFold
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    make_scorer,
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score
)

print("Model training tools imported successfully!")
```

```python
# Step 2: Configure cross-validation and scoring

cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

# Prioritise recall for malignant cases (label 0)
malignant_recall = make_scorer(
    recall_score,
    pos_label=0
)

print("Five-fold cross-validation is ready.")
print("Tuning objective: malignant-class recall")
```

```python
# Step 3: Tune Logistic Regression

lr_params = {
    "model__C": [0.1, 1, 10],
    "model__class_weight": [None, "balanced"]
}

lr_search = GridSearchCV(
    estimator=logistic_pipeline,
    param_grid=lr_params,
    scoring=malignant_recall,
    cv=cv,
    n_jobs=-1,
    refit=True
)

lr_search.fit(X_train, y_train)

best_lr = lr_search.best_estimator_

print("Logistic Regression training completed!")
print("Best parameters:", lr_search.best_params_)
print("Best cross-validation malignant recall:",
      round(lr_search.best_score_, 4))
```

```python
# Step 4: Tune Random Forest

rf_pipeline = Pipeline([
    ("model", RandomForestClassifier(
        random_state=42,
        n_jobs=1
    ))
])

rf_params = {
    "model__n_estimators": [100, 200],
    "model__max_depth": [None, 10],
    "model__class_weight": [None, "balanced"]
}

rf_search = GridSearchCV(
    estimator=rf_pipeline,
    param_grid=rf_params,
    scoring=malignant_recall,
    cv=cv,
    n_jobs=-1,
    refit=True
)

rf_search.fit(X_train, y_train)

best_rf = rf_search.best_estimator_

print("Random Forest training completed!")
print("Best parameters:", rf_search.best_params_)
print("Best cross-validation malignant recall:",
      round(rf_search.best_score_, 4))
```

```python
# Step 5: Evaluate both models on validation data

models = {
    "Logistic Regression": best_lr,
    "Random Forest": best_rf
}

validation_results = []

for name, model in models.items():
    predictions = model.predict(X_val)
    probabilities = model.predict_proba(X_val)

    # Find the probability column corresponding to malignant (label 0)
    malignant_index = list(model.classes_).index(0)
    malignant_prob = probabilities[:, malignant_index]

    results = {
        "Model": name,
        "Accuracy": accuracy_score(y_val, predictions),
        "Malignant Precision": precision_score(
            y_val, predictions, pos_label=0, zero_division=0
        ),
        "Malignant Recall": recall_score(
            y_val, predictions, pos_label=0, zero_division=0
        ),
        "Malignant F1": f1_score(
            y_val, predictions, pos_label=0, zero_division=0
        ),
        "Malignant ROC-AUC": roc_auc_score(
            (y_val == 0).astype(int), malignant_prob
        )
    }

    validation_results.append(results)

validation_df = pd.DataFrame(validation_results)

display(validation_df.round(4))
```

```python
# Step 6: Save validation comparison

validation_df.to_csv(
    "validation_model_comparison.csv",
    index=False
)

print("Validation results saved successfully!")
```

```python
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay

# Store the trained models
models = {
    "Logistic Regression": best_lr,
    "Random Forest": best_rf
}

# Plot confusion matrices using validation predictions
for name, model in models.items():
    predictions = model.predict(X_val)

    cm = confusion_matrix(
        y_val,
        predictions,
        labels=[0, 1]
    )

    display_matrix = ConfusionMatrixDisplay(
        confusion_matrix=cm,
        display_labels=["Malignant", "Benign"]
    )

    display_matrix.plot(cmap="Blues", values_format="d")
    plt.title(f"{name} - Validation Confusion Matrix")
    plt.show()
```

```python
# Step 2: Choose thresholds using validation data only

from sklearn.metrics import fbeta_score

thresholds = np.arange(0.10, 0.91, 0.01)
selected_thresholds = {}

for name, model in models.items():
    # Probability that each case is malignant (label 0)
    malignant_index = list(model.classes_).index(0)
    val_prob = model.predict_proba(X_val)[:, malignant_index]

    # Treat malignant as the positive class for evaluation
    y_val_malignant = (y_val == 0).astype(int)

    scores = []

    for threshold in thresholds:
        val_pred_malignant = (val_prob >= threshold).astype(int)

        score = fbeta_score(
            y_val_malignant,
            val_pred_malignant,
            beta=2,
            zero_division=0
        )

        scores.append(score)

    best_index = int(np.argmax(scores))
    best_threshold = float(thresholds[best_index])

    selected_thresholds[name] = best_threshold

    print(f"\n{name}")
    print("Selected threshold:", round(best_threshold, 2))
    print("Validation F2-score:", round(scores[best_index], 4))
```

```python
# Step 3: Final test-set evaluation

from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay

test_results = []
test_predictions = {}
test_probabilities = {}

for name, model in models.items():
    # Probability of malignant class (label 0)
    malignant_index = list(model.classes_).index(0)
    malignant_prob = model.predict_proba(X_test)[:, malignant_index]

    # Apply the threshold selected on validation data
    threshold = selected_thresholds[name]
    pred_malignant = (malignant_prob >= threshold).astype(int)

    # Convert back to original labels: 0 malignant, 1 benign
    predictions = np.where(pred_malignant == 1, 0, 1)

    test_predictions[name] = predictions
    test_probabilities[name] = malignant_prob

    # Metrics treating malignant as the positive class
    y_test_malignant = (y_test == 0).astype(int)

    accuracy = accuracy_score(y_test, predictions)
    precision = precision_score(
        y_test_malignant, pred_malignant, zero_division=0
    )
    recall = recall_score(
        y_test_malignant, pred_malignant, zero_division=0
    )
    f1 = f1_score(
        y_test_malignant, pred_malignant, zero_division=0
    )
    auc = roc_auc_score(y_test_malignant, malignant_prob)

    test_results.append({
        "Model": name,
        "Threshold": threshold,
        "Accuracy": accuracy,
        "Malignant Precision": precision,
        "Malignant Recall": recall,
        "Malignant F1": f1,
        "ROC-AUC": auc
    })

    # Confusion matrix: original labels 0=malignant, 1=benign
    cm = confusion_matrix(y_test, predictions, labels=[0, 1])

    disp = ConfusionMatrixDisplay(
        confusion_matrix=cm,
        display_labels=["Malignant", "Benign"]
    )

    disp.plot(cmap="Blues", values_format="d")
    plt.title(f"{name} - Test Confusion Matrix")
    plt.show()

test_df = pd.DataFrame(test_results)

display(test_df.round(4))
```

```python
# Step 4: Plot ROC curves

from sklearn.metrics import roc_curve, auc

plt.figure(figsize=(8, 6))

y_test_malignant = (y_test == 0).astype(int)

for name, malignant_prob in test_probabilities.items():
    fpr, tpr, _ = roc_curve(
        y_test_malignant,
        malignant_prob
    )

    roc_auc = auc(fpr, tpr)

    plt.plot(
        fpr,
        tpr,
        label=f"{name} (AUC = {roc_auc:.3f})"
    )

plt.plot([0, 1], [0, 1], "k--", label="Random classifier")

plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate (Malignant Recall)")
plt.title("ROC Curve - Breast Cancer Detection")
plt.legend()
plt.grid(alpha=0.3)
plt.show()
```

```python
# Step 5: Save the evaluation results

test_df.to_csv(
    "test_model_comparison.csv",
    index=False
)

print("Final test metrics saved.")
print("File: test_model_comparison.csv")
```

Model Evaluation and Comparison

Two machine learning models, Logistic Regression and Random Forest, were trained using the Wisconsin Diagnostic Breast Cancer dataset. Hyperparameters were selected using five-fold stratified cross-validation on the training set.

Classification thresholds were selected using validation data, with the F2-score used to emphasise malignant-class recall. The selected models were then evaluated on the separate test set.

```python
# Step 1: Compare models using validation data

from sklearn.metrics import fbeta_score

validation_model_scores = []

for name, model in models.items():
    malignant_index = list(model.classes_).index(0)
    malignant_prob = model.predict_proba(X_val)[:, malignant_index]

    threshold = selected_thresholds[name]
    pred_malignant = (malignant_prob >= threshold).astype(int)
    actual_malignant = (y_val == 0).astype(int)

    validation_model_scores.append({
        "Model": name,
        "Threshold": threshold,
        "Malignant Recall": recall_score(
            actual_malignant, pred_malignant, zero_division=0
        ),
        "Malignant F2": fbeta_score(
            actual_malignant, pred_malignant,
            beta=2, zero_division=0
        ),
        "ROC-AUC": roc_auc_score(
            actual_malignant, malignant_prob
        )
    })

validation_selection_df = pd.DataFrame(validation_model_scores)
display(validation_selection_df.round(4))
```

```python
# Step 2: Select and save the model

import joblib
from pathlib import Path

# Select using validation F2-score, not test results
best_row = validation_selection_df.sort_values(
    by=["Malignant F2", "Malignant Recall", "ROC-AUC"],
    ascending=False
).iloc[0]

best_model_name = best_row["Model"]
final_model = models[best_model_name]
final_threshold = float(selected_thresholds[best_model_name])

# Create a folder for saved models
Path("models").mkdir(exist_ok=True)

# Save the model and its required configuration
model_bundle = {
    "model": final_model,
    "model_name": best_model_name,
    "threshold": final_threshold,
    "feature_names": list(X.columns),
    "label_mapping": {
        0: "Malignant",
        1: "Benign"
    }
}

model_path = "models/best_model.joblib"
joblib.dump(model_bundle, model_path)

print("Selected model:", best_model_name)
print("Selected threshold:", round(final_threshold, 3))
print("Saved to:", model_path)
```

```python
# Step 3: Load and test the saved model

loaded_bundle = joblib.load("models/best_model.joblib")

loaded_model = loaded_bundle["model"]
loaded_threshold = loaded_bundle["threshold"]

sample = X_test.iloc[[0]]

malignant_index = list(loaded_model.classes_).index(0)
malignant_probability = loaded_model.predict_proba(sample)[0][malignant_index]

if malignant_probability >= loaded_threshold:
    prediction = "Malignant"
else:
    prediction = "Benign"

print("Model:", loaded_bundle["model_name"])
print("Prediction:", prediction)
print(
    "Estimated malignant probability:",
    f"{malignant_probability:.2%}"
)
print("Saved model loaded and tested successfully!")
```

```python
%%writefile app.py

import streamlit as st
import pandas as pd
import numpy as np
import joblib
from sklearn.datasets import load_breast_cancer

# Page configuration
st.set_page_config(
    page_title="Breast Cancer Detection",
    page_icon="🔬",
    layout="wide"
)

# Load saved model
bundle = joblib.load("models/best_model.joblib")
model = bundle["model"]
threshold = bundle["threshold"]
model_name = bundle["model_name"]
feature_names = bundle["feature_names"]

# Load dataset to obtain default input values
dataset = load_breast_cancer()
data = pd.DataFrame(
    dataset.data,
    columns=dataset.feature_names
)
default_values = data.median()

# App heading
st.title("🔬 Breast Cancer Detection System")
st.write(
    "Machine learning application using the "
    "Wisconsin Diagnostic Breast Cancer dataset."
)

st.info(
    "Educational project only. This application is not "
    "a medical diagnostic tool. Consult a qualified "
    "healthcare professional for medical decisions."
)

st.write("**Selected model:**", model_name)
st.write("**Decision threshold:**", round(threshold, 3))

st.subheader("Enter Cell Measurements")
st.write(
    "Enter the 30 numerical measurements below. "
    "Default values are dataset medians."
)

# Input fields
input_values = {}

with st.form("prediction_form"):
    columns = st.columns(3)

    for i, feature in enumerate(feature_names):
        with columns[i % 3]:
            input_values[feature] = st.number_input(
                label=feature,
                min_value=0.0,
                value=float(default_values[feature]),
                step=0.1,
                format="%.4f"
            )

    submitted = st.form_submit_button(
        "Predict",
        type="primary",
        use_container_width=True
    )

# Prediction
if submitted:
    input_df = pd.DataFrame(
        [[input_values[f] for f in feature_names]],
        columns=feature_names
    )

    malignant_index = list(model.classes_).index(0)

    probabilities = model.predict_proba(input_df)[0]
    malignant_probability = probabilities[malignant_index]

    if malignant_probability >= threshold:
        prediction = "Malignant"
        st.error("Model prediction: Malignant")
    else:
        prediction = "Benign"
        st.success("Model prediction: Benign")

    st.subheader("Prediction Results")

    col1, col2 = st.columns(2)

    with col1:
        st.metric(
            "Estimated malignant probability",
            f"{malignant_probability * 100:.2f}%"
        )

    with col2:
        st.metric(
            "Estimated benign probability",
            f"{(1 - malignant_probability) * 100:.2f}%"
        )

    st.write("**Predicted class:**", prediction)

    st.caption(
        "These are model-generated probabilities, not "
        "confirmed medical results. The threshold was "
        "selected using validation data."
    )
```
